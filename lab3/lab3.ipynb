### 6. Preparing the data for a binary classification

We will do a multiclass classification on this dataset later. For now, we will start with a binary classification. Update your training and testing sets to contain only examples from two classes: "s" ("Sugi" forest) and  "d" ("Mixed deciduous" forest).

```python
updated_df_train = df_train.loc[df_train['class'].isin(['s','d'])]
print(updated_df_train['class'].value_counts())
x_train = updated_df_train.drop(columns=['class'])
y_train = updated_df_train['class']

print()

updated_df_test = df_test.loc[df_test['class'].isin(['s','d'])]
print(updated_df_test['class'].value_counts())
x_test = updated_df_test.drop(columns=['class'])
y_test = updated_df_test['class']

print(x_train)
print(y_train)
```

### 6.1. Split training data into two sets: training & Validation

Split your training data into two sets: training and validation. Select randomly 80% for training
and 20% for validation. You must set the random generator seed to 42 before the splitting. Ensure
your validation set is fixed and does not change each time you run your algorithm

```python
from sklearn.model_selection import train_test_split, GridSearchCV

x_train, x_train_val, y_train, y_train_val = train_test_split(x_train, y_train, test_size=0.2, random_state=42)
```

### 7. Using the SGDClassifier with Grid Search CV

Use sklearn.linear model.SGDClassifier together with
sklearn.model selection.GridSearchCV to find an optimal value of the regularisation
hyperparameter C and the learning rate η in the training set using Grid Search and a 10-fold crossvalidation. Be careful how scikit-learn defines the Grid Search.

Show the optimal values for C and η according to Grid Search.

```python
from sklearn.linear_model import SGDClassifier
from sklearn.model_selection import KFold

sgd = SGDClassifier(loss='log_loss', learning_rate='constant', random_state=42, max_iter=1000)
param_grid = {
    'eta0': np.logspace(-4, 1, 6), 
    'alpha': np.logspace(-2, 2, 5)
}
grid_search = GridSearchCV(estimator=sgd, param_grid=param_grid, cv = KFold(n_splits=10, shuffle=True, random_state=42), refit=True)
grid_search.fit(x_train,y_train)
print(grid_search.best_params_)
print(grid_search.best_estimator_)
print(grid_search.best_score_)
print(grid_search.best_index_)

best_model = grid_search.best_estimator_
```

Train your model on the entire training set using these optimal values. Using the validation set,
plot precision versus recall and comments on the results. How does the performance measure
behave? What threshold would you choose and why?

```python
from sklearn.metrics import precision_recall_curve

d_idx = list(grid_search.classes_).index('d')
y_scores = grid_search.predict_proba(x_train_val)[:,d_idx]
precision, recall, thresholds = precision_recall_curve(y_train_val, y_scores, pos_label='d')
# print(f'precision: {precision}, recall: {recall}, threshold:{thresholds}')
f1_scores = 2 * (precision * recall) / (precision + recall + np.finfo(float).eps)
idx = np.argmax(f1_scores)
optimal_threshold = thresholds[idx]
print(f'optimal threshold: {optimal_threshold:.3f}')
print(f'optimal F1: {f1_scores[idx]:.3f} (precision: {precision[idx]:.3f}, recall: {recall[idx]:.3f})')

# plot precision vs. recall
plt.figure(figsize=(8, 6))
plt.plot(recall, precision)
plt.plot(recall[idx], precision[idx], 'o', label='optimal threshold')
plt.title('Validation set')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.legend()
plt.show()
```

```python
# accuracy Values
trainScore = best_model.score(x_train, y_train)
testScore = best_model.score(x_test, y_test)

print(f'Accuracy - train {trainScore * 100:.2f}%, test {testScore * 100:.2f}%')
```

```python
y_pred_train = best_model.predict(x_train)
y_pred_test = best_model.predict(x_test)
```

```python
# Confusion Matrix
from sklearn import metrics

cm_train= metrics.confusion_matrix(y_train, y_pred_train)
print("Train cm:\n", cm_train)

cm_test = metrics.confusion_matrix(y_test, y_pred_test)
print("Train cm:\n", cm_test)
```

```python
class_name = ['s','d']
fig, ax = plt.subplots()
im = ax.imshow(cm_train, cmap='plasma', alpha=0.3)
ax.set_xticks([0, 1])
ax.set_xticklabels(['s', 'd'])
ax.set_yticks([0, 1])
ax.set_yticklabels(['s', 'd'])
ax.set_xlabel('Predicted')
ax.set_ylabel('True')
ax.set_title('Confusion Matrix - Train')

for i in range(cm_train.shape[0]):
    for j in range(cm_train.shape[1]):
        ax.text(j, i, cm_train[i, j], ha='center', va='center', color='black')

plt.colorbar(im, ax=ax)
plt.show()
```

```python
disp_test = metrics.ConfusionMatrixDisplay(cm_test, display_labels=class_name)
disp_test.plot(cmap='Blues_r', values_format='d')
plt.title('Confusion Matrix - test')
plt.show()
```

```python
from sklearn.metrics import precision_score, recall_score

d_idx = list(grid_search.classes_).index('d')
y_scores = grid_search.predict_proba(x_train)[:,d_idx]
precision, recall, thresholds = precision_recall_curve(y_train, y_scores, pos_label='d')

# calculate optimal threshold precision and recall on train set
# (use threshold from validation set)
y_train_optim_thresh_pred = (grid_search.predict_proba(x_train)[:, 0] >= optimal_threshold).astype(int)
optim_precision = precision_score(y_train=='d', y_train_optim_thresh_pred)
optim_recall = recall_score(y_train=='d', y_train_optim_thresh_pred)

# plot precision vs. recall
plt.figure(figsize=(8, 6))
plt.plot(recall, precision)
plt.plot(optim_recall, optim_precision, 'o', label='optimal threshold')
plt.title('Training set')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.legend()
plt.show()
```
